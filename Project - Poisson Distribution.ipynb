# Story 1 Heights of Students

```python
import numpy as np
np.random.seed(42)
mean_height = 170  # cm
std_dev = 10       # cm
sample_size = 1000
heights = np.random.normal(loc=mean_height, scale=std_dev, size=sample_size)
sample_mean = np.mean(heights)
sample_variance = np.var(heights)  
print(f"Sample Mean: {sample_mean:.2f} cm")
print(f"Sample Variance: {sample_variance:.2f} cm^2")
```

# Story 1 Poisson Distribution

```python
import numpy as np
np.random.seed(42)
lambda_value = 170 
sample_size = 1000
data = np.random.poisson(lam=lambda_value, size=sample_size)
sample_mean = np.mean(data)
sample_variance = np.var(data)
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Sample Variance: {sample_variance:.2f}")
```

# Story 1 Interpretation

Sample Mean:
The calculated sample mean should be close to the theoretical mean of 170 cm. Small deviations are expected due to random sampling.

Sample Variance: 
The theoretical variance of the distribution is 10^2=100cm^2

Conclusion:
If the sample mean and variance are near the theoretical values, it indicates that the generated data accurately represents a normal distribution with the given parameters. As the sample size increases, these estimates would converge even more closely to the true mean and variance (Law of Large Numbers).

# Story 1 Data Visulization

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
mean_height = 170  
sample_size = 1000
poisson_data = np.random.poisson(lam=mean_height, size=sample_size)
plt.figure(figsize=(8, 5))
plt.hist(poisson_data, bins=30, density=True, color='lightblue', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.title('Histogram of Poisson Distribution (λ = 170)')
plt.show()
```

# Story 2 Daily Temperatures

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
lambda_temp = 5    
sample_size = 300
temperature_data = np.random.poisson(lam=lambda_temp, size=sample_size)
sample_mean = np.mean(temperature_data)
sample_variance = np.var(temperature_data)
print(f"Sample Mean: {sample_mean:.2f} °C")
print(f"Sample Variance: {sample_variance:.2f} (°C²)")
plt.figure(figsize=(8, 5))
plt.hist(temperature_data, bins=15, density=True, color='lightcoral', edgecolor='black')
plt.xlabel('Temperature Value')
plt.ylabel('Probability Density')
plt.title('Histogram of Poisson-Distributed Winter Temperatures (λ = 5)')
plt.show()
```

# Story 3 Call Center Inbound Calls

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
lambda_calls = 3  
sample_size = 500
calls = np.random.poisson(lam=lambda_calls, size=sample_size)
sample_mean = np.mean(calls)
sample_variance = np.var(calls)
print(f"Sample Mean: {sample_mean:.2f} calls per 10 minutes")
print(f"Sample Variance: {sample_variance:.2f} calls²")
plt.figure(figsize=(8, 5))
plt.hist(calls, bins=range(0, max(calls) + 2), density=True,
         color='skyblue', edgecolor='black', align='left')
plt.xlabel('Number of Calls per 10 Minutes')
plt.ylabel('Probability')
plt.title('Poisson Distribution of Inbound Calls (λ = 3)')
plt.show()
```

# Story 3 Interpretation

# Interpretation for Call Center Operations

Mean (~3 calls)
The sample mean should be close to 3, confirming the expected average call volume every 10 minutes.

Variance (~3 calls²)
For a Poisson process, the variance equals the mean, indicating natural randomness in call arrivals.

Operational Significance:

Most 10-minute intervals will have 2–4 calls.

Occasional spikes (e.g., 6–7 calls) are normal and should be planned for.

Staffing decisions can rely on λ = 3 to balance agent availability with expected variability.

# Story 4 Typos in Book Chapter

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
lambda_typos = 2 
sample_size = 2000
typos = np.random.poisson(lam=lambda_typos, size=sample_size)
sample_mean = np.mean(typos)
sample_variance = np.var(typos)
print(f"Sample Mean: {sample_mean:.2f} typos per 50 pages")
print(f"Sample Variance: {sample_variance:.2f} typos²")
plt.figure(figsize=(8, 5))
plt.hist(typos, bins=range(0, max(typos) + 2), density=True,
         color='lightgreen', edgecolor='black', align='left')
plt.xlabel('Number of Typos per 50 Pages')
plt.ylabel('Probability')
plt.title('Poisson Distribution of Typos in a Book Chapter (λ = 2)')
plt.show()
```

# Story 4 Interpretation

# Sample Mean (~2 typos)
The sample mean should be close to 2, confirming the editor’s observation of average typo frequency.

Sample Variance (~2 typos²)
For a Poisson distribution, the mean equals the variance, reflecting the natural randomness of typos.

Implications for the Editor’s Review Process:

Most 50-page sections will contain 1–3 typos.

Sections with 0 typos are possible but less frequent.

Occasional chapters with 4 or more typos are normal and not necessarily a sign of poor writing.

Knowing this variability helps the editor allocate review time efficiently and set realistic quality expectations.
