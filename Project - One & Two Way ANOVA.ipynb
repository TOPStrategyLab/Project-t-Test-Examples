### Story 1 (One-Way ANOVA)

Evaluating the Effectiveness of Three Study Techniques Suppose a researcher wants to test the effect of three different study techniques on student performance in a math exam: group study, self-study, and tutoring. Assume that the students were randomly assigned to one of the study techniques and that the groups are equivalent in terms of prior math knowledge and study habits. The researcher wants to determine if there is a significant difference in the exam scores between the three study techniques. Since there are three groups, the researcher uses One-Way ANOVA to test if the mean exam scores differ significantly between the three study techniques.

1. Set up the hypothesis statements

Null Hypothesis : All study techniques have the same mean exam score.

Alternative Hypothesis:At least one group mean is different.

2. Calculate the p-value and F-statistic with Python, 3. Interpret the results and make conclusions, 4. Visualize the ANOVA test

```python
import numpy as np
from scipy import stats
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
group1 = [85, 90, 92, 88, 91, 85, 90, 92, 88, 91]
group2 = [75, 78, 77, 74, 80, 75, 78, 77, 74, 80]
group3 = [95, 98, 97, 94, 96, 95, 98, 97, 94, 96]
f_statistic, p_value = stats.f_oneway(group1, group2, group3)
print("F-Statistic:", f_statistic)
print("P-Value:", p_value)
```

```python
3. Interpret the results and make conclusions
```

F-statistic is 201.1698 whihc is Very large and the p-value  is 6.033894575409731e-17 which  Much smaller than 0.05

We there  reject the null hypothesis and conclude tht there is a statistically significant difference in mean exam scores among the three study techniques.

```python
4. Visualize the ANOVA test
```

```python
grp1 = pd.DataFrame({'group':'Group Study','value':group1})
grp2 = pd.DataFrame({'group':'Self-Study','value':group2})
grp3 = pd.DataFrame({'group':'Tutoring','value':group3})
forplot = pd.concat([grp1,grp2,grp3])
sns.boxplot(x='group', y='value', data=forplot)
plt.title('The Boxplot for the One-Way ANOVA')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(group1, color='blue', label='Group Study', fill=True)
sns.kdeplot(group2, color='green', label='Self-Study', fill=True)
sns.kdeplot(group3, color='red', label='Tutoring', fill=True)
plt.title('The KDE Plot for the One-Way ANOVA')
plt.legend()
plt.show()
```

The One-Way ANOVA test shows a statistically significant difference between the three study techniques (p < 0.05). Tutoring is the most effective method, followed by group study, while self-study resulted in the lowest average performance.

Story 2 (One-Way ANOVA): Evaluating the Effectiveness of Diet Programs
Suppose a researcher wants to test the effect of three different diets on weight loss: Diet A, Diet B,
and Diet C. The researcher randomly assigns 15 participants to each diet group, ensuring that the groups
are equivalent in terms of age and initial weight. After one month, the researcher measures the amount of
weight lost by each participant. The data is as follows:
• Diet A: [3, 4, 5, 2, 3, 4, 6, 7, 5, 4, 3, 6, 5, 4, 3]
• Diet B: [1, 2, 2, 3, 1, 1, 2, 2, 3, 3, 1, 2, 1, 2, 2]
• Diet C: [6, 7, 8, 7, 6, 8, 9, 8, 7, 9, 8, 7, 6, 7, 8]
The researcher wants to determine if there is a significant difference in weight loss between the three diets.
Since there are three diet groups, the researcher uses a One-Way ANOVA to test if the mean weight loss
differs significantly between the diets. Please answer the following questions:


1. State the null and alternative hypotheses for this One-Way ANOVA test.

Null Hypothesis: There is no difference in mean weight loss among the three diets.

Alternative Hypothesis: At least one diet has a different mean weight loss.

2. How would the researcher interpret the results if the p-value obtained is less than 0.05?

If the p-value is less than the reseracher should reject the null hypothesis and conclude thta there is a statistically significant difference in average weight loss between at least two diets.

3. Calculate the p-value and F-statistic with Python,

```python
import numpy as np
from scipy import stats
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
group1 = [3, 4, 5, 2, 3, 4, 6, 7, 5, 4, 3, 6, 5, 4, 3]  # Diet A
group2 = [1, 2, 2, 3, 1, 1, 2, 2, 3, 3, 1, 2, 1, 2, 2]  # Diet B
group3 = [6, 7, 8, 7, 6, 8, 9, 8, 7, 9, 8, 7, 6, 7, 8]  # Diet C
f_statistic, p_value = stats.f_oneway(group1, group2, group3)
print("F-Statistic:", f_statistic)
print("P-Value:", p_value)
```

4. Interpret the results and make conclusions

Since the pvalue (9.793052354200617e-17) is veru small p<0.05 we reject the null hypothesis and conclude that there is a statistically significant difference in weight loss between the three diets.

5. Visualize the ANOVA test with Python

```python
grp1 = pd.DataFrame({'group':'Diet A','value':group1})
grp2 = pd.DataFrame({'group':'Diet B','value':group2})
grp3 = pd.DataFrame({'group':'Diet C','value':group3})
forplot = pd.concat([grp1,grp2,grp3])
sns.boxplot(x='group', y='value', data=forplot)
plt.title('The Boxplot for the One-Way ANOVA')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(group1, color='blue', label='Diet A', fill=True)
sns.kdeplot(group2, color='green', label='Diet B', fill=True)
sns.kdeplot(group3, color='red', label='Diet C', fill=True)

plt.title('The KDE Plot for the One-Way ANOVA')
plt.legend()
plt.show()
```

The One-Way ANOVA test shows a statistically significant difference in weight loss among the three diets with diet C is the most effective diet program, followed by Diet A, while Diet B is the least effective.

Story 3 (Two-Way ANOVA): Evaluating the Two Factors on Employee Productivity
A researcher is studying the effects of two factors on employee productivity: Training Type and Work
Hours. The two training types are Online Training (OT) and In-Person Training (IP), and the two work
hour levels are 6 hours and 8 hours. The researcher collects productivity scores for 10 employees from
each combination of factors, resulting in the following data:
• Online Training, 6 hours: [50, 55, 52, 58, 54, 60, 53, 59, 61, 56]
• Online Training, 8 hours: [70, 74, 75, 72, 78, 80, 76, 77, 79, 81]
• In-Person Training, 6 hours: [45, 50, 48, 53, 49, 51, 50, 52, 47, 55]
• In-Person Training, 8 hours: [65, 70, 72, 68, 74, 77, 73, 75, 78, 79]
Please answer the following questions:

1. State the null and alternative hypotheses,

H₀₁: Mean productivity is the same for Online and In-Person training

H₁₁: Mean productivity differs by training type

H₀₂: Mean productivity is the same for 6 hours and 8 hours

H₁₂: Mean productivity differs by work hours

H₀₃: There is no interaction between training type and work hours

H₁₃: There is an interaction effect between training type and work hours

2. Check the assumptions before conducting the Two-Way ANOVA,

Employees are independently sampled in the data
Productivity scores within each group should be approximately normal.
Variance across th groups should be approximately equal.

3. Use Python to solve the question

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
import seaborn as sns
import matplotlib.pyplot as plt
data = {
'Training': ['Online']*10 + ['Online']*10 + ['InPerson']*10 + ['InPerson']*10,
'Hours': ['6hrs']*10 + ['8hrs']*10 + ['6hrs']*10 + ['8hrs']*10,
'Productivity': 
[50, 55, 52, 58, 54, 60, 53, 59, 61, 56] +  
[70, 74, 75, 72, 78, 80, 76, 77, 79, 81] +  
[45, 50, 48, 53, 49, 51, 50, 52, 47, 55] +   
[65, 70, 72, 68, 74, 77, 73, 75, 78, 79]     
}
df = pd.DataFrame(data)
model = ols('Productivity ~ C(Training) + C(Hours) + C(Training):C(Hours)', data=df).fit()
anova_results = anova_lm(model)
print(anova_results)
```

4. Interpret the results and make conclusions,

Since p < 0.05, we reject H₀₁ and conclude thta

Hypothesis 2 

F(1,36) = 347.49

p = 4.435809e-20

Since p < 0.05, we reject H₀₂ and conlude thta Work hours have a very strong significant effect on productivity.

Employees working 8 hours are significantly more productive than those working 6 hours.


Hypothesis 3
F(1,36) = 1.34

p = 0.255

Since p > 0.05, we fail to reject H₀₃.


There is no significant interaction between training type and work hours. This means that the he effect of work hours is the same across both training types.


5. Visualize the ANOVA test with Python

```python
sns.pointplot(x='Training', y='Productivity', hue='Hours', data=df)
plt.title("Two-Way ANOVA Interaction Plot")
plt.show()
```

You are conducting an experiment to examine the effect of two factors on the performance of students:
Teaching Method and Study Time. The two teaching methods are Method A and Method B, and the study
times are 2 hours and 4 hours. The goal is to analyze how the teaching method and study time (both

below:

H₁₂: Mean scores differ by study time

You are working as a data analyst for a company that manufactures three different types of smartwatches.
The company wants to understand if there are significant differences in user satisfaction scores across
three factors:
• Battery Life (measured on a scale of 1 to 10),
• Display Quality (measured on a scale of 1 to 10),
• Price (measured on a scale of $100 to $500).
Your task is to simulate data for three different smartwatch models (A, B, and C), each with a sample
size of 50 customers, and then perform a Multivariate Analysis of Variance (MANOVA) to see if these
three variables (Battery Life, Display Quality, and Price) differ significantly across the three models.
Please do the follow tasks with Python.
1. Generate 50 data points from a normal distribution according to the following mean and standard
deviation:
• Smartwatch Model A:
– Battery Life: Mean = 7.5, Standard Deviation = 1.2
– Display Quality: Mean = 8.0, Standard Deviation = 1.0
– Price: Mean = 250, Standard Deviation = 50
• Smartwatch Model B:
– Battery Life: Mean = 6.8, Standard Deviation = 1.3
– Display Quality: Mean = 7.5, Standard Deviation = 1.1
– Price: Mean = 220, Standard Deviation = 60
• Smartwatch Model C:
– Battery Life: Mean = 8.0, Standard Deviation = 1.1
– Display Quality: Mean = 8.5, Standard Deviation = 0.9
– Price: Mean = 300, Standard Deviation = 70

```python
import numpy as np
import pandas as pd
from statsmodels.multivariate.manova import MANOVA
np.random.seed(123) 
n = 50
A_battery = np.random.normal(7.5, 1.2, n)
A_display = np.random.normal(8.0, 1.0, n)
A_price = np.random.normal(250, 50, n)
B_battery = np.random.normal(6.8, 1.3, n)
B_display = np.random.normal(7.5, 1.1, n)
B_price = np.random.normal(220, 60, n)
C_battery = np.random.normal(8.0, 1.1, n)
C_display = np.random.normal(8.5, 0.9, n)
C_price = np.random.normal(300, 70, n)
data = {
'Model': ['A']*n + ['B']*n + ['C']*n,
'Battery_Life': np.concatenate([A_battery, B_battery, C_battery]),
'Display_Quality': np.concatenate([A_display, B_display, C_display]),
'Price': np.concatenate([A_price, B_price, C_price])
}
df = pd.DataFrame(data)
```

2. Fit the MANOVA model

```python
manova = MANOVA.from_formula('Battery_Life + Display_Quality + Price ~ Model', data=df)
result = manova.mv_test()
print(result)
```

```python
3. Interpret the results
```

The MANOVA results indicate a statistically significant difference among the three smartwatch models when battery life, display quality, and price are analyzed simultaneously (Wilks’ Λ = 0.568, F = 15.80, p < 0.001). Therefore we reject the null hypothesis and conclude there is a statistically significant multivariate difference between smartwatch models in terms of battery life, display quality, and price considered together.

Therefore, we reject the null hypothesis and conclude that the overall satisfaction characteristics differ across models. This suggests that model design and pricing strategy significantly influence customer evaluations.
