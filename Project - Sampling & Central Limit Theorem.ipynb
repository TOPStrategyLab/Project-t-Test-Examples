# Story 1 - Random Sampling for a Skewed Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew
np.random.seed(42)
mean_spending = 500
population_size = 10000
population_spending = np.random.exponential(
    scale=mean_spending,
    size=population_size
)
skewness_value = skew(population_spending)
plt.figure(figsize=(10, 6))
plt.hist(population_spending, bins=60, color='skyblue', edgecolor='black')
plt.axvline(np.mean(population_spending), color='red', linestyle='--', label='Mean')
plt.axvline(np.median(population_spending), color='green', linestyle='--', label='Median')
plt.title("Distribution of Monthly Household Spending (Exponential)")
plt.xlabel("Monthly Spending")
plt.ylabel("Frequency")
plt.legend()
plt.text(
    0.65 * max(population_spending),
    max(plt.gca().get_ylim()) * 0.8,
    f"Skewness = {skewness_value:.2f}",
    fontsize=12
)
plt.show()
```

# Story 1 - Interpretation

# Because spending is skewed, relying only on the mean might not be the best. Python users should consider medians, percentiles, or transformations when analyzing the household expenditures.

# Story 2 - Exploring the Law of Large Numbers with Exponential Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
epsilon = 2              
true_mean = 1 / epsilon
n_samples = 1000
samples = np.random.exponential(scale=1/epsilon, size=n_samples)
sample_sizes = np.arange(1, n_samples + 1)
sample_means = np.cumsum(samples) / sample_sizes
plt.figure(figsize=(10, 6))
plt.plot(sample_sizes, sample_means, label="Sample Mean", color="blue")
plt.axhline(true_mean, color="red", linestyle="--", label="True Mean (μ = 0.5)")
plt.xlabel("Sample Size")
plt.ylabel("Sample Mean")
plt.title("Convergence of Sample Mean to True Mean (Exponential Distribution)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

# Story 2 - Interpretation

# Small samples can give misleading averages, especially when the data are skewed. Large samples produce more reliable and stable estimates of the true average. Even when data are not normally distributed, the average still becomes accurate as the sample size grows.This example shows why we prefer larger sample sizes when working with real-world data that are uneven or skewed.

# Story 3 - Finishing a Training Module

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
theta = 2     
varpi = 5    
n_population = 300
n_samples = 40
sample_size = 25
min_time = 10
max_time = 60
beta_data = np.random.beta(theta, varpi, size=n_population)
completion_times = min_time + beta_data * (max_time - min_time)
sample_means = []
for _ in range(n_samples):
    sample = np.random.choice(completion_times, size=sample_size, replace=False)
    sample_means.append(np.mean(sample))
sample_means = np.array(sample_means)
plt.figure(figsize=(9, 6))
plt.hist(sample_means, bins=10, color="lightcoral", edgecolor="black")
plt.axvline(np.mean(sample_means), color="blue", linestyle="--", label="Average of Sample Means")
plt.xlabel("Sample Mean Completion Time (minutes)")
plt.ylabel("Frequency")
plt.title("Histogram of Sample Means (40 samples of size 25)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

# Story 3 - Interpretation

# The original completion times are heavily skewed because most employees either finish very quickly or take a long time, with fewer employees in the middle. This makes the raw data uneven and lopsided.However, when we take many samples and calculate the average completion time for each sample, the distribution of those averages looks more bell-shaped.

# Story 4 - Central Limit Theorem with Uniform Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
lower = 0
upper = 10
true_mean = (lower + upper) / 2 
n = 1000
weights = np.random.uniform(lower, upper, n)
sample_sizes = np.arange(1, n + 1)
sample_means = np.cumsum(weights) / sample_sizes
plt.figure(figsize=(10, 6))
plt.plot(sample_sizes, sample_means, label="Sample Mean", color="blue")
plt.axhline(true_mean, color="red", linestyle="--", label="True Mean (5 grams)")
plt.xlabel("Sample Size")
plt.ylabel("Sample Mean (grams)")
plt.title("Convergence of Sample Mean to True Mean\nUniform Distribution (0–10 grams)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
fixed_sample_size = 50
num_simulations = 1000
means = []
for _ in range(num_simulations):
    sample = np.random.uniform(lower, upper, fixed_sample_size)
    means.append(np.mean(sample))
plt.figure(figsize=(9, 6))
plt.hist(means, bins=30, color="lightgreen", edgecolor="black")
plt.axvline(true_mean, color="red", linestyle="--", label="True Mean (5 grams)")
plt.xlabel("Sample Mean (grams)")
plt.ylabel("Frequency")
plt.title("Distribution of Sample Means (n = 50)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

# Story 4 - Interpretation

# Each individual product weight is completely random between 0 and 10 grams, so the raw data are spread out evenly with no clear “peak.” Early on, when only a few products are measured, the average weight jumps around a lot and can be far from 5 grams.As more products are included in the sample, the average becomes more stable and steadily moves closer to 5 grams, which is the true average weight of all products. By the time hundreds of products are included, the sample mean barely changes at all. When we look at many sample averages taken from larger samples, those averages form a bell-shaped distribution, even though the original product weights are uniformly distributed.

# Story 5 - Central Limit Theorem with Normal Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
mu = 2          
variance = 1
sigma = np.sqrt(variance)
n = 1000
weights = np.random.normal(loc=mu, scale=sigma, size=n)
sample_sizes = np.arange(1, n + 1)
sample_means = np.cumsum(weights) / sample_sizes
plt.figure(figsize=(10, 6))
plt.plot(sample_sizes, sample_means, label="Sample Mean", color="blue")
plt.axhline(mu, color="red", linestyle="--", label="True Mean (2 grams)")
plt.xlabel("Sample Size")
plt.ylabel("Sample Mean (grams)")
plt.title("Convergence of Sample Mean to True Mean\nNormal Distribution (µ = 2, σ² = 1)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

# Story 5 - Interpretation

# Each individual product weight varies slightly around the target of 2 grams. When only a few products are measured, the average weight can bounce around and may not reflect the true target weight very well. As more products are included in the sample, the average weight becomes much more stable. The influence of unusually light or heavy products gets balanced out, and the sample mean settles close to 2 grams. Because the product weights already follow a normal distribution, the sample mean behaves very predictably and converges smoothly to the true mean as the sample size increases.

# Story 6 - Central Limit Theorem with Exponential Distribution in Customer Service

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
epsilon = 1         
true_mean = 1 / epsilon
n = 1000
waiting_times = np.random.exponential(scale=1/epsilon, size=n)
sample_sizes = np.arange(1, n + 1)
sample_means = np.cumsum(waiting_times) / sample_sizes
plt.figure(figsize=(10, 6))
plt.plot(sample_sizes, sample_means, label="Sample Mean Waiting Time", color="blue")
plt.axhline(true_mean, color="red", linestyle="--", label="Expected Mean (1 minute)")
plt.xlabel("Sample Size")
plt.ylabel("Average Time Between Calls (minutes)")
plt.title("Convergence of Sample Mean for Exponential Waiting Times")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

# Story 6 - Interpretation

# The time between customer calls is uneven. Most calls arrive quickly, but sometimes there are longer gaps, which makes the data right-skewed. Because of this, the average waiting time calculated from just a few calls can vary a lot. When the sample size is small, the sample mean jumps around and may not be close to the true average of 1 minute. This happens because a single long gap between calls can strongly affect the average. As more call times are included, the effect of unusually long waiting times is reduced. The sample mean becomes more stable and moves closer to 1 minute. By the time hundreds of call intervals are included, the average hardly changes at all. Even though the original waiting times are skewed, the behavior of the sample mean becomes smooth and predictable as the sample size increases.
