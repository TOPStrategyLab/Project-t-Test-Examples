## Story 1: Average Hours on Homework (One-Sample Two Sided T-test)

## Story 1: Hypothesis Statement

#### The school principal wants to know whether students are spending an average of 10 hours per week on homework, as expected, or if that number is actually different. To examine this belief, I set up two opposing statements. The null hypothesis assumes that the average amount of time students spend on homework each week is exactly 10 hours. In other words, any difference we see in the sample is simply due to random chance or normal variation, not an actual change in the average.The alternative hypothesis suggests that the average amount of time students spend on homework each week is not 10 hours. This means the true average could be either higher or lower than 10 hours, indicating a real difference from what the principal believes.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
data = np.array([9, 12, 10, 11, 8, 14, 9, 11, 12, 13, 8, 10, 9, 11, 10, 12, 9, 10, 8, 11])
mu_0 = 10
t_stat, p_value = stats.ttest_1samp(data, mu_0)
print("T-statistic:", round(t_stat, 3))
print("P-value:", round(p_value, 3))
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The average homework hours is significantly different from 10.")
else:
    print("Fail to reject the null hypothesis: No significant difference from 10 hours.")
mean = np.mean(data)
std = np.std(data, ddof=1)
x = np.linspace(mean - 4*std, mean + 4*std, 1000)
y = stats.norm.pdf(x, mean, std)
plt.figure(figsize=(8, 5))
plt.plot(x, y, color='teal', label='Sample distribution')
plt.axvline(mu_0, color='red', linestyle='--', label='Hypothesized Mean (10 hrs)')
plt.axvline(mean, color='blue', linestyle='-', label=f'Sample Mean ({mean:.2f} hrs)')
plt.title('Bell Curve of Weekly Homework Hours')
plt.xlabel('Homework Hours per Week')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

## Story 1: Interpretation

#### I collected data from 20 students to investigate the school principal’s belief that the average number of hours students spend on homework each week is different from 10 hours. Using a two-tailed one-sample t-test, the sample mean of approximately 10.55 hours per week was compared to the hypothesized mean of 10 hours.The statistical test produced a t-statistic and a p-value greater than the significance level of 0.05. Since the p-value exceeds 0.05, there is no sufficient evidence to reject the null hypothesis. The data do not show a meaningful difference between the students’ average homework time and the hypothesized average of 10 hours per week. Thus, based on this sample, it appears reasonable to conclude that students, on average, spend about 10 hours per week on homework, in line with the principal’s belief.

## Story 1: Type 1 & Type 2 Error Explained

#### Although I have drawn the conclusion based on statistical evidence, there are always potential risks of error in hypothesis testing. A Type I error, also known as a false alarm, occurs when the null hypothesis is mistakenly rejected, the null is actually true. In this study, that would mean concluding that students’ average homework time differs from 10 hours when it truly does not. The likelihood of making this type of mistake is represented by α (alpha), which I set at 5%, or a 0.05 probability. On the other hand, a Type II error, or missed detection, happens when we fail to reject the null hypothesis even though it is false. In this case, this would mean concluding that students spend about 10 hours per week on homework when, in fact, the real average is different, perhaps 9 or 11 hours. The probability of making this kind of error is called β (beta), which depends on factors such as the sample size, the variability in the data, and how large the true difference from the hypothesized mean actually is.

## Story 2: Average Product Weight (One-Sample One Sided T-Test)

## Story 2: Hypothesis Statement

#### The manufacturer claims that the average weight of their product is greater than 500 grams. To evaluate this claim, I compare the sample data to the standard of 500 grams. The null hypothesis assumes that the average product weight is equal to or less than 500 grams. This means any observed increase in average weight in the sample is simply due to random variation. The alternative hypothesis contends that the average product weight is greater than 500 grams, supporting the manufacturer’s claim. I am testing whether the data provides enough evidence to show that the products are, on average, truly heavier than 500 grams.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
data = np.array([505, 510, 490, 520, 515, 530, 495, 510, 500, 505, 525, 515, 490, 520, 505])
mu_0 = 500
t_stat, p_value_two_tailed = stats.ttest_1samp(data, mu_0)
if t_stat > 0:
    p_value = p_value_two_tailed / 2
else:
    p_value = 1 - (p_value_two_tailed / 2)
print("Sample Mean:", round(np.mean(data), 2))
print("T-statistic:", round(t_stat, 3))
print("One-tailed P-value:", round(p_value, 3))
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The average product weight is significantly greater than 500 grams.")
else:
    print("Fail to reject the null hypothesis: No evidence that the average product weight is greater than 500 grams.")
mean = np.mean(data)
std = np.std(data, ddof=1)
x = np.linspace(mean - 4*std, mean + 4*std, 1000)
y = stats.norm.pdf(x, mean, std)
plt.figure(figsize=(8, 5))
plt.plot(x, y, color='teal', label='Sample Distribution')
plt.axvline(mu_0, color='red', linestyle='--', label='Hypothesized Mean (500g)')
plt.axvline(mean, color='blue', linestyle='-', label=f'Sample Mean ({mean:.2f}g)')
plt.title('Bell Curve of Product Weights')
plt.xlabel('Weight (grams)')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

## Story 2: Intepretation

#### After analyzing the sample data, the average product weight was approximately 510 grams, slightly higher than the claimed standard of 500 grams. When conducting the one-tailed t-test at a 5% significance level, the p-value (probability of observing such a result by random chance) is less than 0.05, which allows me to reject the null hypothesis. This means that the data provides strong statistical evidence to support the manufacturer’s claim that their products weigh more than 500 grams on average. If the p-value had been greater than 0.05, I would have concluded that there isn’t enough evidence to say the products are heavier.

## Story 2: Type 1 & Type 2 Error Explained

#### A Type I error would occur if we conclude that the products are heavier than 500 grams. In this context, we would be accepting the manufacturer’s claim when it’s actually false. The probability of making this type of error is represented by α (alpha), which we set at 5% (0.05). A Type II error, on the other hand, would happen if we fail to detect that the products are indeed heavier — in other words, concluding there’s no difference when the true average weight really is greater than 500 grams. The probability of this missed detection is denoted as β (beta), which depends on the sample size, the variability of the data, and how much the true mean exceeds 500 grams.

## Story 3: Adjusting the Sample Size

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
x = np.linspace(-10, 5, 1000)
std_dev = 2
mean_h0 = 0
mean_h1 = -2
y_h0 = norm.pdf(x, mean_h0, std_dev)
y_h1 = norm.pdf(x, mean_h1, std_dev)
plt.plot(x, y_h0, label="Null Hypothesis (No Weight Loss)", linestyle="--")
plt.plot(x, y_h1, label="Alternative Hypothesis (Weight Loss)")
plt.title("Bell Curve Distributions for Weight-Loss Study")
plt.xlabel("Average Weight Change (lbs)")
plt.ylabel("Probability Density")
plt.legend()
plt.show()
```

## Story 3: Type 2 Error Explanation

#### In this study, a Type II error would occur if the weight-loss program actually works but I fail to detect the effect and conclude that there is no impact. When I only include 10 participants, individual differences and random weight changes can easily mask a real but small amount of weight loss, making this error more likely. By increasing the sample size, I reduce the influence of random variation and get a clearer estimate of the program’s true effect. This makes my study more powerful and lowers the chance that I will miss a real weight-loss effect, thereby reducing the probability of making a Type II error.

## Story 3: Type 1 Error Explanation (Increasing the Sample Size)

#### Increasing the sample size would not affect the probability of making a Type I error as long as I keep the same significance level in my study. A Type I error happens when I conclude that the weight-loss program works even though it actually does not, and that risk is controlled by the alpha level I choose (such as 0.05), not by how many participants I include. While a larger sample helps me detect real effects more easily, it does not increase or decrease the chance of a false positive unless I change the significance level.

## Story 3: Detecting True Effect

#### If the true effect of the program is small, a larger sample size is especially important because small changes in weight can easily be hidden by normal day-to-day fluctuations and differences between people. When I use only a few participants, random variation can make it look like the program has no effect even if it does help a little. By increasing the sample size, I can average out that random noise and get a more precise estimate of the program’s impact. This makes it easier for me to detect a small but real weight-loss effect that I might otherwise miss.

## Story 4: Why the t-Distribution is More Important than the z-Score

#### I use the t-distribution instead of the z-score because my sample size is very small (only six students) and I do not know the true population standard deviation. With such a small group, my estimate of variability is less precise, and the t-distribution accounts for that extra uncertainty. The z-score assumes I know the population standard deviation and have a large sample, which is not true in this case.

## Story 4: How the t-Distribution Adjust for Uncertainty with Small Samples

#### The t-distribution adjusts for uncertainty by being wider and having heavier tails than the standard normal distribution. This means I need stronger evidence before concluding that my sample mean is significantly different from the national average. By spreading the curve out more, the t-distribution reflects the fact that results from a small sample are more variable and less reliable.

## Story 4: What Happens as the Sample Size Increases? 

#### As my sample size increases, the t-distribution gradually becomes narrower and starts to look more like the standard normal distribution. This happens because my estimate of the standard deviation becomes more accurate with more data. Eventually, when the sample size is large, the difference between using a t-distribution and a z-distribution becomes very small.
